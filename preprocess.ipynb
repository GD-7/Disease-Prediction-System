# Preprocessing

Import

```python
import pandas as pd
```

Load Dataset

```python
data = pd.read_csv('data/dataset.csv')

data.head()
```

Preprocess

```python
symptom_columns = data.columns[1:]

# Replace NaN
data[symptom_columns] = data[symptom_columns].fillna("None")

# Get all unique symptoms
all_symptoms = set()
for col in symptom_columns:
    all_symptoms.update(data[col].unique())

all_symptoms.discard("None")

# Create empty dataframe for new features
encoded_df = pd.DataFrame(0, index=data.index, columns=list(all_symptoms))

# Fill values
for col in symptom_columns:
    for symptom in all_symptoms:
        encoded_df.loc[data[col] == symptom, symptom] = 1

# Combine with Disease column
final_df = pd.concat([encoded_df, data[['Disease']]], axis=1)

# Remove duplicates
final_df.drop_duplicates(inplace=True)

final_df.head()
```

Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
final_df['Disease'] = le.fit_transform(final_df['Disease'])

mapping_df = pd.DataFrame({
    'Disease': le.classes_,
    'Encoded_Value': le.transform(le.classes_)
})

print(mapping_df)
```

Save preprocessed dataset

```python
final_df['Disease'].value_counts()
final_df.to_csv('data/preprocessed_dataset.csv', index=False)
```

Save encoding mapping

```python
mapping_df.to_csv("data/disease_encoding_map.csv", index=False)
```
